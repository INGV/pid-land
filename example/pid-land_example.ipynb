# PID-LAND Example Notebook (Simplified)

This notebook simulates the PID-LAND workflow: resolution of PIDs, extraction of WF-Handle and WF-Provenance, and generation of WF-Manifest.
All operations are performed in-memory using hard-coded JSONs, with no dependency on real PID-LAND services or external files.

```python
# PID-LAND Example Notebook (Simulated)

import json
from pathlib import Path
import pandas as pd
```

```python
# -----------------------------
# Hard-coded RUM output dataset
# -----------------------------
rum_dataset = {
    "prefix/my-test-handle-PID": {
        "WF-Handle": {
            "@context": {
                "dc": "http://purl.org/dc/elements/1.1/",
                "dcterms": "http://purl.org/dc/terms/",
                "schema": "http://schema.org/",
                "file": "http://schema.org/DigitalDocument"
            },
            "dc_identifier": "prefix/my-test-handle-PID",
            "dc_title": "Example mSEED Repository",
            "dc_creator": "Example Institution",
            "dc_type": "Dataset, seismic waveform",
            "dcterms_spatial": {
                "x": "LON",
                "y": "LAT",
                "z": "ALT"
            }
        },
        "WF-Provenance": {
            "@context": {
                "dc": "http://purl.org/dc/elements/1.1/",
                "dcterms": "http://purl.org/dc/terms/",
                "schema": "http://schema.org/",
                "file": "http://schema.org/DigitalDocument"
            },
            "dc_identifier": "prefix/my-test-handle-PID",
            "dcterms:isPartOf": "wf_handle",
            "prov_SoftwareAgent": "https://github.com/obspy/obspy",
            "schema_file": {
                "position": "http://hdl.handle.net/prefix/my-test-handle-PID",
                "name": "/content/sample_data/anscombe.json"
            }
        }
    }
}
```

# Simulated resolver functions

```python
# -----------------------------
# Simulated resolver functions
# -----------------------------
def resolve_pid(pid, urlappend=None):
    """
    Simulate PID-LAND resolver for a given PID.
    urlappend can be 'metadata', 'provenance', or None.
    """
    if pid not in rum_dataset:
        raise ValueError(f"PID {pid} not found in dataset")

    if urlappend == "metadata":
        return rum_dataset[pid]["WF-Handle"]
    elif urlappend == "provenance":
        return rum_dataset[pid]["WF-Provenance"]
    else:
        # Default: download data
        raw_file = Path(rum_dataset[pid]["WF-Provenance"]["schema_file"]["name"]) # very fake data file
        df = pd.read_csv(raw_file)
        
        return df.head()
```

# Example usage

```python
pid = "prefix/my-test-handle-PID"

print("Resolve metadata (WF-Handle):")
print(json.dumps(resolve_pid(pid, urlappend="metadata"), indent=2))

print("\nResolve provenance (WF-Provenance):")
print(json.dumps(resolve_pid(pid, urlappend="provenance"), indent=2))

print("\nDownload data")
print(resolve_pid(pid))
```
